```python
df_clean = df.copy()

df_clean['published_at'] = pd.to_datetime(df_clean['published_at'])
df_clean['fetch_date'] = pd.to_datetime(df_clean['fetch_date'])

print(df_clean[['published_at', 'fetch_date']].dtypes)
```

```python
df_clean['title'] = df_clean['title'].str.strip()
df_clean['channel'] = df_clean['channel'].str.strip()

print((df_clean['title'] != df_clean['title'].str.strip()).sum())
print((df_clean['channel'] != df_clean['channel'].str.strip()).sum())
```

```python
print(df_clean['title'].str.contains('\n').sum())
print(df_clean['title'].str.contains('  ').sum())
print(df_clean['title'].str.contains('\u200b').sum())
```

```python
df_clean['title'] = df_clean['title'].str.replace('\u200b', '', regex=False)
df_clean['title'] = df_clean['title'].str.replace(r'\s+', ' ', regex=True)
df_clean['title'] = df_clean['title'].str.strip()

print(df_clean['title'].str.contains('\n').sum())
print(df_clean['title'].str.contains('  ').sum())
print(df_clean['title'].str.contains('\u200b').sum())
```

```python
print(df_clean['channel'].str.contains('\n').sum())
print(df_clean['channel'].str.contains('  ').sum())
print(df_clean['channel'].str.contains('\u200b').sum())
```

```python
df_clean['channel'] = df_clean['channel'].str.replace(r'\s+', ' ', regex=True)
print(df_clean['channel'].str.contains('  ').sum())
```

```python
cols = ['views', 'likes', 'comments']
df_clean[cols] = df_clean[cols].replace(0, np.nan)
print(df_clean[cols].isnull().sum())
```

```python
print(df_clean['country'].nunique())
print(df_clean['country'].unique())
```

```python
print(df_clean['channel'].nunique())
print(df_clean['channel'].str.lower().nunique())
```

```python
names = df_clean['channel'].drop_duplicates()
dup = names[names.str.lower().duplicated(keep=False)]
print(sorted(dup))
```

```python
df_clean = df_clean.rename(columns={'fetch_date': 'trending_date'})
print(df_clean.columns.tolist())
```

```python
order = ['video_id', 'title', 'channel', 'published_at',
         'country', 'trending_date',
         'views', 'likes', 'comments']
df_clean = df_clean[order]
print(df_clean.columns.tolist())
```

```python
print('Строк до:', len(df), '| после:', len(df_clean))
print('Полных дубликатов:', df_clean.duplicated().sum())
print('Повторов видео+страна+дата:', df_clean.duplicated(subset=['video_id', 'country', 'trending_date']).sum())
print('Отрицательных чисел:', (df_clean[['views', 'likes', 'comments']] < 0).sum().sum())
print('ID не из 11 символов:', (df_clean['video_id'].str.len() != 11).sum())
print('Пробелы по краям в названиях:', (df_clean['title'] != df_clean['title'].str.strip()).sum())
print('Публикация позже трендов:', (df_clean['published_at'].dt.tz_localize(None) > df_clean['trending_date']).sum())
print(df_clean.dtypes)
```

```python
print('Rows before:', len(df), '| after:', len(df_clean))
print('Full duplicates:', df_clean.duplicated().sum())
print('Duplicate video+country+date:', df_clean.duplicated(subset=['video_id', 'country', 'trending_date']).sum())
print('Negative values:', (df_clean[['views', 'likes', 'comments']] < 0).sum().sum())
print('Invalid video_id length:', (df_clean['video_id'].str.len() != 11).sum())
print('Leading/trailing spaces in titles:', (df_clean['title'] != df_clean['title'].str.strip()).sum())
print('Published after trending:', (df_clean['published_at'].dt.tz_localize(None) > df_clean['trending_date']).sum())
print(df_clean.dtypes)
```

```python
# 1. Full clean dataset
df_clean.to_csv('/content/drive/MyDrive/youtube_trending_clean.csv', index=False)

# 2. Excel sample: US only
df_us = df_clean[df_clean['country'] == 'US'].copy()
df_us['published_at'] = df_us['published_at'].dt.tz_localize(None)
df_us = df_us.rename(columns={'published_at': 'published_at_utc'})
df_us.to_excel('/content/drive/MyDrive/youtube_trending_clean_US.xlsx', index=False)

print('CSV rows:', len(df_clean))
print('Excel (US) rows:', len(df_us))
```
